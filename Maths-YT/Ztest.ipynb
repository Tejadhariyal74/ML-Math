```python
import numpy as np
from scipy.stats import norm
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]    #your sample size must be greater than 30. for using z-score
population_mean = 170
population_std= 3                             # standard deviation given hona chhiye population ka.
sample_mean= np.mean(sample)
n = len(sample)
```

```python
z_score = (sample_mean - population_mean)/(population_std/np.sqrt(n))
z_score
```

```python
p_value = 2 * (1- norm.cdf(abs(z_score)))
p_value
```

```python
alpha = 0.05

if p_value < alpha:
    print("I will reject the null hypothesis")

else:
    print("I will accept the null hypothesis")
```
