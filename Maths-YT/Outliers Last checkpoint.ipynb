#### Statistics code implementation

```python
import numpy as np
arr= np.array([2,3,4,6,7,8,9,12,13,16,17,23,25,27,34,37,201])
```

```python
arr
```

```python
Q1= np.percentile(arr,25)           # 25 percentile-> Q1
# Q1                                   # 50 percentile-> Q2
Q3= np.percentile(arr,75)           # 25 percentile-> Q1
# Q3                                  # 75 percentile-> Q3
```

```python
IQR= Q3-Q1       
IQR
# ab upperfance and lowerfance niklenge because upperfance and lowerfance ke bahar outliers hote hai.
```

```python
UF=Q3+ (1.5*IQR)
UF
```

```python
LF=Q3-(1.5*IQR) 
LF
```

```python
l= []
for i in arr:
    if i >= LF and i <= UF:
        l.append(i)

l
# we have removed our outlier.
```

```python
# now converting list to arr
arr2= np.array(l)
```

```python
print(arr)
print(arr2)
```

```python
# now builting box plot
import seaborn as sns
sns.boxplot(x=arr)   # arr because arr mai outliers hai.
sns.boxplot(x=arr2)  # arr2 because arr2 mai outliers nahi hai. 
```
