##### code implementation of two sample t-test

```python
import numpy as np
from scipy import stats
```

```python
group_A = [85, 88, 90, 92, 87, 85, 89, 91, 86, 88]
group_B = [82, 84, 80, 83, 81, 79, 78, 85, 84, 83]

# null- hypothesis -> there is no diffrence between these two sample. dono same terike se work kr rhhe hia.
# alternate- hypothesis -> there is a diffrence between group Aand group B.
```

```python
t_stats , p_value = stats.ttest_ind(group_A, group_B, equal_var= False)   # always write false in equal_vari when applying t-test
```

```python
t_stats
```

```python
alpha = 0.05
if p_value < alpha:
    print("I will reject the null hypothesis")
else:
    print("I will accept the null hypothesis")
```
